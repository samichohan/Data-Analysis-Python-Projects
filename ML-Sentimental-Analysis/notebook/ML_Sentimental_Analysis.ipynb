# 2. Remove HTML Tags

```python
import re
def remove_html(text):
  clean = re.compile('<.*?>')
  return re.sub(clean,'',text)
```

# 3.Remove URL'S

```python
def remove_url(text):
  url=re.compile(r'https?://\S+|www\.\S+')
  return url.sub(r'',text)
```

# 4.Remove Punctuations

```python
import string
string.punctuation
```

```python
def remove_punctuations(text):
  punctuations = string.punctuation
  no_punctuations="".join([char for char in text if char not in punctuations])
  return no_punctuations
```

```python
#Split dataset into train and test
from sklearn.model_selection import train_test_split
x_train, x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=100)
```

# Classification

1.NaiveBayes

2.Logistic Regression

3.Support Vector Machine

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
```

```python
svc = SVC(kernel='sigmoid',gamma=1.0)
mnb = MultinomialNB()
lrc = LogisticRegression(solver='liblinear',penalty='l1')
```

```python
classifiers = {
    'SVC':svc,
    'NB':mnb,
    'LR':lrc
}
```

```python
from sklearn.metrics import accuracy_score, precision_score, confusion_matrix, recall_score,f1_score
```

```python
def train_classifier(classifiers,x_train,y_train,x_test,y_test):
  classifiers.fit(x_train,y_train)
  y_pred = classifiers.predict(x_test)
  accuracy=accuracy_score(y_test,y_pred)
  precision=precision_score(y_test,y_pred)
  recall=recall_score(y_test,y_pred)
  f1 = f1_score(y_test,y_pred)
  confusion=confusion_matrix(y_test,y_pred)

  return accuracy,precision,recall,f1,confusion
```

# Model Performance And Evaluation

```python
accuracy_scores = []
precision_scores = []
recall_scores = []
f1_scores = []
for name,clf in classifiers.items():
  current_accuracy,current_precision,current_recall,current_f1,current_confusion=train_classifier(clf,x_train,y_train,x_test,y_test)

  print('Name:',name)
  print('Accuracy:',current_accuracy)
  print('Precision:',current_precision)
  print('Recall:',current_recall)
  print('F1:',current_f1)
  print('Confusion:',current_confusion)

  accuracy_scores.append(current_accuracy)
  precision_scores.append(current_precision)
  recall_scores.append(current_recall)
  f1_scores.append(current_f1)
```

```python
performanced_df=pd.DataFrame({'Algorithm':classifiers.items(),'Accuracy':accuracy_scores,'Precision':precision_scores,'Recall':recall_scores,'F1':f1_scores})
```
